```python
import numpy

distance_file = 'distance_data_headers.csv'
distances = numpy.genfromtxt(fname=distance_file, delimiter=',', dtype='unicode')
headers = distances[0]
data = distances[1:]
data = data.astype(float)
```

## Plotting Data

```python
import matplotlib.pyplot as plt
plt.figure()
plt.plot(data[:,1])
```

## Labeling plots and saving figures

```python
sample = headers[1]

plt.figure()
plt.xlabel('Simulation Frame')
plt.ylabel('Distance (angstrom)')
fig_1 = plt.plot(data[:,1], label=sample)
plt.legend()

plt.savefig(F'{sample}.png')
```

## Increasing image resolution

```python
sample = headers[1]

plt.figure()
plt.xlabel('Simulation Frame')
plt.ylabel('Distance (angstrom)')
fig_1 = plt.plot(data[:,1], label=sample)
plt.legend()

plt.savefig(F'{sample}.png', dpi=300)
```

## Plotting more than one set of data

```python
plt.figure()
plt.xlabel('Simulation Frame')
plt.ylabel('Distance (angstrom)')
plt.plot(data[:,1], label=headers[1])
plt.plot(data[:,2], label=headers[2])
plt.legend()
plt.savefig('two_samples.png')
```

```python
for col in range(1, len(headers)):
    fig = plt.plot(data[:,col], label=headers[col])
    plt.legend()

plt.xlabel('Simulation Frame')
plt.ylabel('Distance (angstrom)')
plt.savefig('all_samples.png')
```

## Exercise

```python
# Loop: used enumerate(headers [1:], start=1) to get the column index and sample name at once skipping column 0 (frame numbers).
# Separate plots: Placed plt.figure() inside the loop so each sample opens on a new canvas instead of overlapping. 
for col, sample in enumerate(headers[1:], start=1):
    plt.figure()
    plt.plot(data[:, col], label=sample)
    plt.xlabel('Simulation Frame')
    plt.ylabel('Distance (angstrom)')
    plt.legend()
    plt.savefig(f'{sample}.png')
```

```python
#data[:, 1:]: Grabs all rows (:) and only the distance columns (1:), skipping column 0 since those are just frame numbers and would
#totally ruin the y-axis scale. * 0.9 and * 1.1: Drops the min by 10% and increae the max up by 10% so the lines are not squished right against
#the top and bottom edges of the graph.

data_min = numpy.min(data[:, 1:]) * 0.9
data_max = numpy.max(data[:, 1:]) * 1.1

for col, sample in enumerate(headers[1:], start=1):
    plt.figure()
    plt.plot(data[:, col], label=sample)
    plt.ylim(data_min, data_max)
    plt.xlabel('Simulation Frame')
    plt.ylabel('Distance (angstrom)')
    plt.legend()
    plt.savefig(f'{sample}.png')
```

```python
fig, ax = plt.subplots(len(headers)-1, 1)
for col in range(1, len(headers)):
    sample = headers[col]
   
    ax[col-1].plot(data[0::100,0], data[0::100,col], label=sample)
    ax[col-1].set_xlabel('Simulation Frame')
    ax[col-1].set_ylabel('Distance (angstrom)')
    ax[col-1].legend()
```

```python
fig, ax = plt.subplots(len(headers)-1, 1, figsize=(10, 10), sharey=True)
for col in range(1, len(headers)):
    sample = headers[col]
    ax[col-1].plot(data[0::100, 0], data[0::100, col], label=sample)
    ax[col-1].set_xlabel('Simulation Frame')
    ax[col-1].set_ylabel('Distance (ängstrom)')
    ax[col-1].legend()

fig.tight_layout()
```
